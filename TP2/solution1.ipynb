```python
#Step 1
days = ["Mon","Tues","Wen","Thus","Fri","Satur","Sun"]
temps = [31,33,29,34,35,32,30]
```

```python
#step 2
print("Number of days:",len(days))
print("First days :",days[0],temps[0])
print("Last Days :",days[-1],temps[-1])
```

```python
import numpy as np
```

```python
# step 3
print("The hightest:",max(temps))
print("The Lowest is:",min(temps))
average = sum(temps)/len(temps)
print("AVG is:",average)
```

```python
#Step 4
hottest = temps.index(max(temps))
print("The hottest day is:",days[hottest])
```

```python
#step 4
print("Sorted: ",sorted(temps))
print("Original :",temps)
```
