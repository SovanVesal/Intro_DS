```python
#step 1
grades = {
    "Dara": [80,75,90],
    "Sophea": [95,88,92],
    "Vibol": [40,55,45],
    "Malis": [70,85,78]
}
```

```python
#step 2
average = {}
for name,score in grades.items():
    average[name] = round(sum(score)/len(score),1)
print("Average:",average)
```

```python
#step 3
passed = set()
for name,avg in grades.items():
    if avg >= [50] :
        print(name, "PASS")
        passed.add(name)
    else:
        print(name,"FALSE")
print("Passed:",sorted(passed))
```

```python
#step 4
ranking = sorted(average.items(),key=lambda item: item[1], reverse=True)
for rank ,(name ,avg) in enumerate(ranking,start=1):
    print(f"#{rank} {name} {avg}")
```
